```python
#cross-validation dengan model baseline dummy classifier
cross_val_score(dummy_clf, X_train, y_train_5, cv=3, scoring="accuracy") #mengukur 'accuracy' perfoma model dengan cross-validation
```

```python
#buat cross validation secara manual atau proses latar belakang
from sklearn.model_selection import StratifiedKFold #import StratifiedKFold dari sklearn.model_selection
from sklearn.base import clone #import clone dari sklearn.base  

skfold= StratifiedKFold(n_splits=3, random_state=42, shuffle=True) #buat instance dari StratifiedKFold dengan 3 lipatan, random_state=42 dan shuffle=True

for train_index, test_index in skfold.split(X_train, y_train_5): #lakukan perulangan untuk setiap lipatan
    clone_clf = clone(sgd_clf) #buat salinan dari model sgd_clf
    X_train_folds = X_train[train_index] #ambil data latih untuk lipatan saat ini
    y_train_folds = y_train_5[train_index] #ambil label latih untuk lipatan saat ini
    X_test_fold = X_train[test_index] #ambil data uji untuk lipatan saat ini
    y_test_fold = y_train_5[test_index] #ambil label uji untuk lipatan saat ini

    clone_clf.fit(X_train_folds, y_train_folds) #latih model dengan data latih dan label latih untuk lipatan saat ini
    y_pred = clone_clf.predict(X_test_fold) #prediksi label untuk data uji pada lipatan saat ini
    n_correct = sum(y_pred == y_test_fold) #hitung jumlah prediksi yang benar pada lipatan saat ini
    print(n_correct / len(y_pred)) #hitung akurasi pada lipatan saat ini
```

```python
# Membuat Precision-Recall Curve untuk melihat performa model, Fungsi ini mencoba berbagai threshold dan menghitung precision dan recall untuk setiap threshold. Hasilnya adalah tiga array: precision, recall, dan thresholds. Precision dan recall adalah metrik yang digunakan untuk mengevaluasi performa model klasifikasi, sedangkan thresholds adalah nilai ambang yang digunakan untuk menentukan apakah suatu prediksi dianggap positif atau negatif. Jadi secara konsep kita mendapatkan tabel
from sklearn.metrics import precision_recall_curve #import precision_recall_curve dari sklearn.metrics

precisions, recalls, thresholds = precision_recall_curve(y_train_5, y_scores) #buat precision_recall_curve dengan data latih dan score dari decision function
```

tabel precision, recall dan threshold yang dibuat

```python

import pandas as pd

pr_curve_df = pd.DataFrame({
    "threshold": thresholds,
    "precision": precisions[:-1],
    "recall": recalls[:-1]
})

print(pr_curve_df)
```

```python
#buat plot precision-recall curve untuk melihat performa model versi sederhana
plt.plot(thresholds, precisions[:-1], "b--", label="Precision", linewidth=2)
plt.plot(thresholds, recalls[:-1], "g-", label="Recall", linewidth=2)
plt.vlines(threshold, 0, 1.0, "k", "dotted", label="threshold")
plt.axis([-50000, 50000, 0, 1])
plt.grid()
plt.xlabel("Threshold")
plt.legend(loc="center right")
plt.show()
```

```python
# bikin figure (kanvas gambar) baru, ukuran 8x4 inci
plt.figure(figsize=(8, 4))

# --- ini bagian UTAMA: gambar precision dan recall sebagai fungsi dari threshold ---

# garis putus-putus biru = precision, sumbu x = thresholds, sumbu y = precisions
# precisions[:-1] karena ingat, panjang array precisions itu 1 LEBIH BANYAK dari thresholds
# (di titik terakhir precision=1, recall=0, nggak punya threshold pasangannya)
# jadi elemen terakhir dibuang biar panjangnya sama-sama pas buat di-plot
plt.plot(thresholds, precisions[:-1], "b--", label="Precision", linewidth=2)

# garis solid hijau = recall, dengan alasan pemotongan [:-1] yang sama
plt.plot(thresholds, recalls[:-1], "g-", label="Recall", linewidth=2)

# gambar SATU garis vertikal putus-putus hitam di posisi x = threshold (nilai threshold yang dipilih)
# garisnya membentang dari y=0 sampai y=1.0
# ini nunjukin "di sinilah threshold yang kita pilih (misal 3000)"
plt.vlines(threshold, 0, 1.0, "k", "dotted", label="threshold")

# --- bagian di bawah ini cuma dekorasi/penanda titik, bukan wajib buat plot utamanya ---

# cari index pertama di array thresholds yang nilainya >= threshold yang kita pilih
# (thresholds >= threshold) menghasilkan array boolean [False, False, ..., True, True, ...]
# .argmax() otomatis ambil index kemunculan True yang PERTAMA
idx = (thresholds >= threshold).argmax()

# gambar titik bulat BIRU tepat di posisi (threshold, precision) pada index tsb
# tujuannya nunjukin nilai precision persis di titik threshold yang dipilih
plt.plot(thresholds[idx], precisions[idx], "bo")

# gambar titik bulat HIJAU tepat di posisi (threshold, recall) pada index tsb
# tujuannya nunjukin nilai recall persis di titik threshold yang dipilih
plt.plot(thresholds[idx], recalls[idx], "go")

# --- kembali ke pengaturan grafik standar ---

plt.axis([-50000, 50000, 0, 1])   # atur batas sumbu: x dari -50000 s/d 50000, y dari 0 s/d 1
plt.grid()                         # tampilkan garis grid di background
plt.xlabel("Threshold")            # label sumbu x
plt.legend(loc="center right")     # tampilkan kotak legenda di tengah-kanan

plt.show()  # tampilkan grafiknya
```

```python
#grafik sederhanya Precision/Recall Curve untuk melihat performa model
plt.plot(recalls, precisions, linewidth=2, label="Precision/Recall curve")
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.axis([0, 1, 0, 1])
plt.grid()
plt.legend(loc="lower left")
plt.show()
```

```python
import matplotlib.patches as patches  # modul tambahan buat bikin bentuk panah melengkung

# bikin figure (kanvas gambar) baru dengan ukuran 6x5 inci
plt.figure(figsize=(6, 5))

# ini yang UTAMA: gambar kurva precision-recall
# sumbu x = recalls, sumbu y = precisions, garis dengan ketebalan 2
plt.plot(recalls, precisions, linewidth=2, label="Precision/Recall curve")

# --- bagian di bawah ini cuma dekorasi/penanda, bukan bagian wajib ---

# garis putus-putus VERTIKAL dari sumbu x (y=0) naik sampai ke titik precision pada threshold ke-idx
# tujuannya nunjukin "di titik recall ini, precision-nya berapa"
plt.plot([recalls[idx], recalls[idx]], [0., precisions[idx]], "k:")

# garis putus-putus HORIZONTAL dari sumbu y (x=0) ke titik recall pada threshold ke-idx
# tujuannya nunjukin "di titik precision ini, recall-nya berapa"
plt.plot([0.0, recalls[idx]], [precisions[idx], precisions[idx]], "k:")

# gambar satu titik bulat hitam ("ko" = warna hitam, marker bulat) tepat di posisi (recall, precision)
# titik ini mewakili satu threshold tertentu (di contoh buku: threshold = 3000)
plt.plot([recalls[idx]], [precisions[idx]], "ko",
         label="Point at threshold 3,000")

# gambar panah melengkung dari titik (0.79, 0.60) menuju titik (0.61, 0.78)
# ini cuma elemen visual buat nunjuk ke arah "kalau threshold dinaikkan, titik geser ke arah sini"
plt.gca().add_patch(patches.FancyArrowPatch(
    (0.79, 0.60), (0.61, 0.78),        # titik awal dan titik akhir panah
    connectionstyle="arc3,rad=.2",      # bikin panahnya melengkung, bukan garis lurus
    arrowstyle="Simple, tail_width=1.5, head_width=8, head_length=10",  # bentuk & ukuran panah
    color="#444444"))                   # warna panah abu-abu gelap

# taruh teks label "Higher threshold" di dekat panah, sebagai penjelasan arah panah
plt.text(0.56, 0.62, "Higher\nthreshold", color="#333333")

# --- bagian ini kembali ke pengaturan grafik standar ---

plt.xlabel("Recall")        # label sumbu x
plt.ylabel("Precision")     # label sumbu y
plt.axis([0, 1, 0, 1])      # atur batas sumbu: x dari 0-1, y dari 0-1 (karena precision & recall selalu 0-1)
plt.grid()                  # tampilkan garis grid di background biar enak dibaca

plt.legend(loc="lower left")  # tampilkan kotak legenda (keterangan garis/titik) di pojok kiri bawah

plt.show()  # tampilkan grafiknya
```

ROC CURVE

```python
#buat ROC Curve untuk melihat performa model, Fungsi ini mencoba berbagai threshold dan menghitung true positive rate (TPR) dan false positive rate (FPR) untuk setiap threshold. Hasilnya adalah tiga array: fpr, tpr, dan thresholds. TPR adalah rasio dari jumlah prediksi positif yang benar dibandingkan dengan jumlah total prediksi positif yang sebenarnya, sedangkan FPR adalah rasio dari jumlah prediksi positif yang salah dibandingkan dengan jumlah total prediksi negatif yang sebenarnya. Jadi secara konsep kita mendapatkan tabel
from sklearn.metrics import roc_curve #import roc_curve dari sklearn.metrics

fpr, tpr, thresholds = roc_curve(y_train_5, y_scores) #buat roc_curve dengan data latih dan score dari decision function
```

Preview Tabelnya

```python
#tpr adalah recall, sedangkan fpr adalah false positive rate (nilai negatif yang dianggap positif). Jadi kita bisa membuat dataframe untuk melihat nilai tpr dan fpr pada setiap threshold

import pandas as pd

roc_curve_df = pd.DataFrame({
    "threshold": thresholds,
    "fpr": fpr,
    "tpr": tpr
})

print(roc_curve_df[(roc_curve_df['tpr'] > 0.4) & (roc_curve_df['tpr'] <= 0.5)]) #cek nilai tpr yang lebih besar dari 0.4 sampai 0.5
```

Poin penting yang perlu dipahami:

argmax() pada array boolean akan ambil index True pertama. Ini cuma valid kalau thresholds terurut menurun (dari tinggi ke rendah) — pola umum saat pakai precision_recall_curve atau roc_curve dari sklearn.

Bagian "extra code" (arrow & text) itu murni dekorasi visual, tidak mempengaruhi hasil analisis — bisa dihapus kalau kamu cuma mau lihat kurva ROC-nya polos.

```python
# Mencari index threshold pertama yang nilainya <= threshold_for_90_precision
# (thresholds biasanya urut menurun, jadi argmax() ambil index pertama yang True)
idx_for_threshold_at_90 = (thresholds <= threshold_for_90_precision).argmax()

# Ambil nilai TPR (True Positive Rate) dan FPR (False Positive Rate)
# pada index threshold 90% precision tadi, buat ditandai di grafik
tpr_90, fpr_90 = tpr[idx_for_threshold_at_90], fpr[idx_for_threshold_at_90]

# Bikin figure/canvas baru untuk plot, ukurannya 6x5 inch
plt.figure(figsize=(6, 5))  # extra code – not needed, just formatting

# Plot kurva ROC utama: FPR di sumbu x, TPR di sumbu y
plt.plot(fpr, tpr, linewidth=2, label="ROC curve")

# Plot garis diagonal putus-putus sebagai baseline classifier acak (random guess)
plt.plot([0, 1], [0, 1], 'k:', label="Random classifier's ROC curve")

# Tandai titik (fpr_90, tpr_90) dengan marker bulat hitam ("ko")
# ini titik di kurva ROC yang sesuai dengan threshold 90% precision
plt.plot([fpr_90], [tpr_90], "ko", label="Threshold for 90% precision")

# extra code – just beautifies Figure 3–7
# Menambahkan panah melengkung sebagai anotasi visual di grafik
plt.gca().add_patch(patches.FancyArrowPatch(
    (0.20, 0.89), (0.07, 0.70),          # titik awal & akhir panah
    connectionstyle="arc3,rad=.4",        # bikin panah melengkung
    arrowstyle="Simple, tail_width=1.5, head_width=8, head_length=10",
    color="#444444"))

# Menambahkan teks label "Higher threshold" di dekat panah tadi
plt.text(0.12, 0.71, "Higher\nthreshold", color="#333333")

# Kasih label untuk sumbu x dan y
plt.xlabel('False Positive Rate (Fall-Out)')
plt.ylabel('True Positive Rate (Recall)')

# Nyalakan grid/garis bantu di background plot
plt.grid()

# Set batas sumbu x dan y dari 0 sampai 1
plt.axis([0, 1, 0, 1])

# Tampilkan legend (keterangan garis/titik) di pojok kanan bawah
plt.legend(loc="lower right", fontsize=13)

# Render dan tampilkan plot-nya
plt.show()
```

```python
def plot_error_grid(X_aa, X_ab, X_ba, X_bb, cl_a, cl_b,
                    n=5, pad=0.2, img_shape=(28, 28)):
    """Gambar 4 blok contoh gambar (2x2) untuk analisis error kelas a vs b."""
    fig, ax = plt.subplots(figsize=(n, n))

    step = n + pad          # jarak dari satu blok ke blok berikutnya
    total = 2 * n + pad     # panjang total tiap sumbu

    # posisi blok: (kolom, baris) -> kelompok gambarnya
    # kolom 0 = diprediksi a, kolom 1 = diprediksi b
    # baris 0 = kelas asli b (bawah), baris 1 = kelas asli a (atas)
    blocks = {(0, 0): X_ba, (1, 0): X_bb,
              (0, 1): X_aa, (1, 1): X_ab}

    for (col, row), images in blocks.items():
        for idx, img in enumerate(images[:n * n]):
            x = idx % n + col * step
            y = idx // n + row * step
            ax.imshow(img.reshape(img_shape), cmap="binary",
                      extent=(x, x + 1, y, y + 1))

    # label kelas di tengah tiap blok
    centers = [n / 2, step + n / 2]
    ax.set_xticks(centers, [str(cl_a), str(cl_b)])
    ax.set_yticks(centers, [str(cl_b), str(cl_a)])

    # garis pemisah antar blok
    ax.axvline(n + pad / 2, color="k", linestyle=":")
    ax.axhline(n + pad / 2, color="k", linestyle=":")

    ax.axis([0, total, 0, total])
    ax.set_xlabel("Predicted label")
    ax.set_ylabel("True label")
    plt.show()

# cara pakai
plot_error_grid(X_aa, X_ab, X_ba, X_bb, cl_a, cl_b)
```
